```python
import numpy as np

SEED = 42

rng = np.random.default_rng(SEED)

n_simulations = 100_000

peak_demand = rng.normal(
    loc=4000,
    scale=400,
    size=n_simulations
)

pump_capacity = rng.normal(
    loc=4500,
    scale=300,
    size=n_simulations
)

insufficient_capacity = pump_capacity < peak_demand

probability = insufficient_capacity.mean()

print(f"Probability: {probability:.1%}")
```
